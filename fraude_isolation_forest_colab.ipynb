# Detección de Fraude de Tarjetas (Isolation Forest)

Proyecto de detección de anomalías listo para **Google Colab**.

**Pregunta:** ¿cómo detectar transacciones fraudulentas?

**Método:** Isolation Forest

**Stack:** pandas, scikit-learn, matplotlib

Datos simulados.

## 0. Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
print("OK")
```

## 1. Generar datos simulados

```python
np.random.seed(42)
n_normal = 1000
n_fraude = 50

normal = np.random.normal(0, 1, (n_normal, 2))
fraude = np.random.normal(5, 1, (n_fraude, 2))

X = np.vstack([normal, fraude])
y = np.array([0]*n_normal + [1]*n_fraude)

df = pd.DataFrame(X, columns=["monto", "frecuencia"])
df["fraude"] = y
print(df.shape)
df.head()
```

## 2. Visualización

```python
plt.scatter(df["monto"], df["frecuencia"], c=df["fraude"], cmap="coolwarm", alpha=0.6)
plt.title("Transacciones (rojo = fraude)")
plt.xlabel("Monto")
plt.ylabel("Frecuencia")
plt.show()
```

## 3. Isolation Forest

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[["monto", "frecuencia"]])

iso = IsolationForest(contamination=0.05, random_state=42)
preds = iso.fit_predict(X_scaled)
# -1 = anomalía
df["pred"] = (preds == -1).astype(int)
print(df["pred"].value_counts())
```

## 4. Evaluación

```python
from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(df["fraude"], df["pred"]))
print(confusion_matrix(df["fraude"], df["pred"]))
```

## 5. Conclusiones

- Isolation Forest es efectivo para detectar anomalías sin etiquetas.
- Ideal para fraude cuando el porcentaje de positivos es muy bajo.
- Se puede extender a más features (hora, ubicación, etc.).

**Consejo:** ajustá `contamination` según el ratio esperado de fraude.
